```python
dbutils.widgets.text(
    "input_path",
    "",
    "Input Path"
)

dbutils.widgets.text(
    "batch_id",
    "",
    "Batch ID"
)

dbutils.widgets.dropdown(
    "source_load_type",
    "incremental",
    ["full", "incremental"],
    "Source Load Type"
)

dbutils.widgets.dropdown(
    "execution_mode",
    "standard",
    ["standard", "backfill"],
    "Execution Mode"
)

```

```python
input_path = dbutils.widgets.get("input_path")
batch_id = dbutils.widgets.get("batch_id")
source_load_type = dbutils.widgets.get(
    "source_load_type"
)
execution_mode = dbutils.widgets.get(
    "execution_mode"
)
```

## Calling Bronze Notebook

```python
print("STEP 1: Raw -> Bronze")

dbutils.notebook.run(
    "./01_raw_to_bronze",
    0,
    {
        "input_path": input_path,
        "batch_id": batch_id,
        "source_load_type": source_load_type,
        "execution_mode": execution_mode,
    }
)

print("Raw -> Bronze complete")
```

## Calling Silver

```python
print("STEP 2: Bronze -> Silver")

dbutils.notebook.run(
    "./02_bronze_to_silver",
    0,
    {
        "batch_id": batch_id,
        "execution_mode": execution_mode,
    }
)

print("Bronze -> Silver complete")
```
